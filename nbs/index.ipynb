# Electrical Docs

> Electrical theory, signal processing, load forecasting and energy cost: the maths and data side of electrical engineering.

- skip_showdoc: true
- skip_exec: true


This site covers electricity on paper and in data, not on the bench. It starts with the equations, moves to signal processing, then forecasts grid load from real demand data, and ends by asking whether a home battery pays for itself.

For the hardware side (tools, motors, microcontrollers, switchboards) see [Hardware Tools](https://bthek1.github.io/Hardware_Tools/). For the general forecasting toolkit these pages build on see [Time Series ML](https://bthek1.github.io/TimeSeriesML/).

## Not Covered Yet

- **No network theorems.** Thevenin, Norton and superposition are missing, so circuit analysis stops at Ohm and Kirchhoff.
- **No three-phase or power systems page.** Nothing on power factor correction, transmission, or how the grid balances supply and demand.
- **Control systems is one section of a reference sheet.** No page on transfer functions, Bode plots or PID tuning.
- **No filter design.** The FFT page forecasts with Fourier terms but never designs an analog or digital filter.
- **The forecasting data is American.** Both forecasting pages use PJM and ERCOT data. Nothing yet uses Australian NEM (National Electricity Market) demand or prices.
